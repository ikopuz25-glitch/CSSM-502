```python
import statsmodels.api as sm
```

```python
y = data['age']
X = data['prestg10_num']
```

```python
X
```

```python
X = sm.add_constant(X)
```

```python
X
```

```python
model = sm.OLS(y, X, missing='drop')
model_result = model.fit()
model_result.summary()
```

```python
# This is not a sociologically meaningful equation at all but I could not turn the others' type from category to float thus chose age. 
# Of course I would rather compare prestige of occupational status with income.
# but here, i am trying to guess the age of the respondents with correlation to their occupational prestige. 
```

```python
#The ols model has a significant p-value. R^2 is very very low<1 so IV explains very little about the variance DV. 
#For the coefficients: constant = 44.0500 and prestg10_num= 0.1450, so the beta_0= 44.05, and beta_1=0.145 in the OLS equation where y=age and x=prestg_num
```

```python
from scipy import stats
mu, std = stats.norm.fit(model_result.resid)
mu, std
```

```python
#####this is from the coursenotes

import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots()
sns.histplot(x=model_result.resid, ax=ax, stat="density", linewidth=0, kde=True)
ax.set(title="Distribution of residuals", xlabel="residual")

xmin, xmax = plt.xlim() 
x = np.linspace(xmin, xmax, 100) 
p = stats.norm.pdf(x, mu, std) 
sns.lineplot(x=x, y=p, color="orange", ax=ax)
plt.show()
```

```python
fitted = model_result.fittedvalues
residuals = model_result.resid
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.scatterplot(x=fitted, y=residuals)
plt.axhline(0, color='red', linestyle='--')
```

```python
sns.histplot(residuals, kde=True)
plt.title("Residuals Distribution")
plt.show()
```

```python
sns.boxplot(x=model_result.resid, showmeans=True)
```

```python
from statsmodels.stats.diagnostic import het_breuschpagan

bp_test = het_breuschpagan(residuals, model_result.model.exog)
labels = ['LM Statistic', 'LM-Test p-value', 'F-Statistic', 'F-Test p-value']

dict(zip(labels, bp_test))
```

```python
###p-value<0.05 thus heteroscedasticity, this means homoscedasticity is violated. 
```

```python
###"In regression and time-series modelling, basic forms of models make use of the assumption that the errors or disturbances 
#ε_i have the same variance across all observation points. When this is not the case, the errors are said to be heteroskedastic, 
#or to have heteroskedasticity, and this behaviour will be reflected in the residuals ε_i estimated from a fitted model." 
#I understand this error can be caused by the size of the sample. 
```

```python
# I look at some pages on internet and learned that changing the OLS model would be better for getting rid of this error
# Also I learned one can use robust computing methods for standard errors and ask chatgbt to write me code on that regard
# which worked but I would not say I am completely following at this point, I understand the OLS equation. 

robust_result = model_result.get_robustcov_results(cov_type='HC3')
print(robust_result.summary())
```
