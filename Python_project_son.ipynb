```python
import pandas as pd

rows = []
```

```python
from lxml import etree

NS = {"tei": "http://www.tei-c.org/ns/1.0"}

for file_path in tei_files:
    tree = etree.parse(file_path)
    root = tree.getroot()

    for u in root.xpath(".//tei:u", namespaces=NS):
        text = " ".join(u.xpath(".//text()", namespaces=NS))
        text = clean_whitespace(text)

        if text:
            rows.append({ "file": file_path, "speech": text})
```

```python
df = pd.DataFrame(rows)

df.shape
```

```python
df.head()
```

```python
df["speech_lower"] = df["speech"].str.lower()
```

```python
def contains_keywords(text, keywords):
    for kw in keywords:
        if kw in text:
            return 1
    return 0
```

```python
for category, keywords in ENV_KEYWORDS.items():
    df[category] = df["speech_lower"].apply(
        lambda x: contains_keywords(x, keywords)
    )
```

```python
env_cols = list(ENV_KEYWORDS.keys())

df["env_speech"] = (df[env_cols].sum(axis=1) > 0).astype(int)
```
